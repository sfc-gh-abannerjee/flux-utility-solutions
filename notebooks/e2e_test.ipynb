# Flux E2E Test Notebook

Tests Semantic View and Agent creation from notebook deployment path.

```python
# Get Snowflake session
from snowflake.snowpark.context import get_active_session
session = get_active_session()

# Configuration
DATABASE = 'FLUX_E2E_TEST'
print(f'Testing with database: {DATABASE}')
```

```python
# Test 1: Verify Semantic View exists
result = session.sql(f"SHOW SEMANTIC VIEWS IN SCHEMA {DATABASE}.APPLICATIONS").collect()
if result:
    print(f'✅ Semantic View found: {result[0]["name"]}')
else:
    print('❌ No Semantic View found')
```

```python
# Test 2: Verify Agent exists
result = session.sql(f"SHOW AGENTS IN SCHEMA {DATABASE}.APPLICATIONS").collect()
if result:
    print(f'✅ Agent found: {result[0]["name"]}')
else:
    print('❌ No Agent found')
```

```python
# Test 3: Query data through semantic view tables
result = session.sql(f"""
    SELECT COUNT(*) as customer_count 
    FROM {DATABASE}.PRODUCTION.CUSTOMERS_MASTER_DATA
""").collect()
print(f'✅ Customer count: {result[0][0]}')
```
